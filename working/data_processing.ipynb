## Mood analysis using valence
Group songs into happy (valence > 0.6), neutral, and sad (valence < 0.4) and plot counts.

```python
def classify_mood(valence):
    if valence > 0.6:
        return 'Happy'
    elif valence < 0.4:
        return 'Sad'
    else:
        return 'Neutral'

songs['mood'] = songs['valence'].apply(classify_mood)
```

```python
mood_counts = songs['mood'].value_counts()
mood_counts
```

```python
plt.figure(figsize=(7, 5))
sns.barplot(
    x=mood_counts.index,
    y=mood_counts.values,
    palette=['#FFD166', '#A0AEC0', '#6B7280']
)

plt.xlabel('Mood')
plt.ylabel('Valence')
plt.title('Scatter Plot of Songs by Mood (Valence-based)')

plt.show()
```

## Instrumental vs vocal tracks
Analyze instrumentalness and find how many tracks are likely instrumental.

Predicts whether a track contains no vocals. "Ooh" and "aah" sounds are treated as instrumental in this context. Rap or spoken word tracks are clearly "vocal". The closer the instrumentalness value is to 1.0, the greater likelihood the track contains no vocal content. Values above 0.5 are intended to represent instrumental tracks, but confidence is higher as the value approaches 1.0.

## Speechiness analysis
Identify rap/spoken tracks using speechiness > 0.33 and plot genre distribution.

Speechiness detects the presence of spoken words in a track. The more exclusively speech-like the recording (e.g. talk show, audio book, poetry), the closer to 1.0 the attribute value. Values above 0.66 describe tracks that are probably made entirely of spoken words. Values between 0.33 and 0.66 describe tracks that may contain both music and speech, either in sections or layered, including such cases as rap music. Values below 0.33 most likely represent music and other non-speech-like tracks.

```python
songs['speechiness'].agg(['min', 'max'])
```

```python
rap = (songs['speechiness'].apply(lambda x: 'rap' if x >= 0.33 else 'non-rap' if x < 0.5 else 'Neutral').value_counts())
rap
```

```python
rap_no_rap = songs.groupby('playlist_genre')['speechiness'].agg(rap=lambda x: (x >= 0.33).sum(), not_rap=lambda x: (x < 0.33).sum())
rap_no_rap
```

```python
plt.figure(figsize=(6, 6))
plt.pie(
    rap_no_rap['rap'],
    labels=rap_no_rap.index,
    autopct='%1.1f%%',
    startangle=90
)
plt.title('Distribution of Rap Tracks by Genre')
plt.show()
```
