## Popularity analysis
Analyze distribution of track_popularity. Plot histogram and identify most and least popular tracks.

```python
songs.columns
```

```python
popular_most = songs.groupby(["track_name","track_artist"])["track_popularity"].max().sort_values(ascending=False)
popular_least = (songs.groupby(["track_name","track_artist"])["track_popularity"].max().sort_values(ascending=True))
```

```python
plt.subplot(1,2,1)
plt.hist(popular_most, bins=10, color='blue', edgecolor='black')
plt.title('Most popular tracks')
plt.xlabel('Most popular tracks')
plt.ylabel('Number of tracks')

plt.subplot(1,2,2)
plt.hist(popular_least, bins=10, color='red', edgecolor='white')
plt.title('Least popular tracks')
plt.xlabel('Least popular tracks')
plt.ylabel('Number of tracks')

plt.tight_layout()
plt.show()
```

## Artist popularity ranking
Find top 10 artists based on average track_popularity. Plot a bar chart.

```python
songs["track_artist"].value_counts()
```

```python
artist_popular = (songs.groupby("track_artist")["track_popularity"].mean().sort_values(ascending=False).head(5))

fig, ax = plt.subplots(figsize=(9, 5))
ax.bar(artist_popular.index, artist_popular.values, color='teal')

plt.tight_layout()
plt.show()
```

## Genre vs popularity
Compare average popularity across playlist_genre and playlist_subgenre using bar plots.

```python
playlist_popular = (songs.groupby("playlist_genre")["track_popularity"].mean().sort_values(ascending=False))
plt.figure(figsize=(10, 6))
sns.barplot(x=playlist_popular.values, y=playlist_popular.index, palette='magma')
```

```python
sns.violinplot(data=songs, x='playlist_genre', y='track_popularity')
plt.title('Popularity by Genre')
plt.xlabel('Playlist Genre')
plt.ylabel('Popularity')
plt.xticks(rotation=45)
plt.show()
```

```python
genre_counts = songs['playlist_genre'].value_counts()
plt.figure(figsize=(8, 8))
plt.pie(genre_counts, labels=genre_counts.index, autopct='%1.1f%%', startangle=90, colors=sns.color_palette('pastel'))
plt.title('Genre Distribution in Playlists')
plt.show()
```
